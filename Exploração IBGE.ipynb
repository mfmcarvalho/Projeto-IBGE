#1 Obtendo o catálogo de dados do IBGE para determinar quais tabelas eu quero extrair

```python
import json
import csv
import requests

URL = "https://servicodados.ibge.gov.br/api/v3/agregados"

print("Baixando catálogo do SIDRA...")

response = requests.get(URL, timeout=120)
response.raise_for_status()

catalogo = response.json()

print(f"Total de registros recebidos: {len(catalogo)}")

# JSON completo
with open("sidra_catalogo.json", "w", encoding="utf-8") as f:
    json.dump(catalogo, f, ensure_ascii=False, indent=2)

# Descobrir automaticamente as colunas existentes
if catalogo:
    campos = sorted({
        chave
        for registro in catalogo
        if isinstance(registro, dict)
        for chave in registro.keys()
    })

    with open("sidra_catalogo.csv", "w", encoding="utf-8-sig", newline="") as f:
        writer = csv.DictWriter(
            f,
            fieldnames=campos,
            extrasaction="ignore"
        )

        writer.writeheader()
        writer.writerows(catalogo)

print("Arquivos criados:")
print("  sidra_catalogo.json")
print("  sidra_catalogo.csv")
```
